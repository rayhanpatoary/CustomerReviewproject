```python
df_work = df[
    [
        "name",
        "asins",
        "brand",
        "categories",
        "reviews.rating",
        "reviews.text",
        "reviews.title",
        "reviews.doRecommend"
    ]
].copy()

df_work.columns = [
    "product_name",
    "product_id",
    "brand",
    "category",
    "rating",
    "review_text",
    "review_title",
    "recommended"
]

display(df_work.head())

print("Shape:", df_work.shape)

print("\nMissing values:")
print(df_work.isnull().sum())

print("\nRating distribution:")
print(df_work["rating"].value_counts().sort_index())

print("\nUnique products:")
print(df_work["product_id"].nunique())
```

```python
# Fill missing product names using another row with the same product_id
name_map = (
    df_work.dropna(subset=["product_name"])
    .drop_duplicates("product_id")
    .set_index("product_id")["product_name"]
)
```

```python
df_work["product_name"] = df_work["product_name"].fillna(
    df_work["product_id"].map(name_map)
)
```

```python
# Remove rows where essential information is missing
df_work = df_work.dropna(
    subset=["product_id", "rating", "review_text"]
)
```

```python
print("Shape after cleaning:", df_work.shape)

print("\nMissing values:")
print(df_work.isnull().sum())

print("\nRating distribution:")
print(df_work["rating"].value_counts().sort_index())

print("\nUnique products:")
print(df_work["product_id"].nunique())
```

```python
# Remove duplicate reviews
df_work = df_work.drop_duplicates(
    subset=["product_id", "review_text"]
)
```

```python
def create_sentiment(rating):
    if rating <= 2:
        return "Negative"
    elif rating == 3:
        return "Neutral"
    else:
        return "Positive"
```

```python

print("\nRating distribution:")
print(df_work["rating"].value_counts().sort_index())

print("\nUnique products:")
print(df_work["product_id"].nunique())
```

```python
df_work["sentiment"] = df_work["rating"].apply(create_sentiment)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

baseline_model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            max_features=20000,
            ngram_range=(1, 2),
            stop_words="english"
        )
    ),
    (
        "classifier",
        LogisticRegression(
            max_iter=1000,
            class_weight="balanced"
        )
    )
])

baseline_model.fit(
    train_df["review_text"],
    train_df["sentiment"]
)
```

```python
from sklearn.metrics import classification_report, accuracy_score

val_predictions = baseline_model.predict(
    val_df["review_text"]
)
print(val_predictions)
```

```python
print(
    "Validation Accuracy:",
    accuracy_score(
        val_df["sentiment"],
        val_predictions
    )
)
```

```python
negative_train = train_df[train_df["sentiment"] == "Negative"]
neutral_train = train_df[train_df["sentiment"] == "Neutral"]
positive_train = train_df[train_df["sentiment"] == "Positive"]

n = len(negative_train)
print("Negative:", n)

n = len(neutral_train)
print("Neutral:", n)

n = len(positive_train)
print("Positive:", n)
```

```python
balanced_train_df = pd.concat([
    negative_train.sample(n=n, random_state=42),
    neutral_train.sample(n=n, random_state=42),
    positive_train.sample(n=n, random_state=42)
])

balanced_train_df = balanced_train_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

print(balanced_train_df["sentiment"].value_counts())
print("Total:", len(balanced_train_df))
```

```python
label2id = {
    "Negative": 0,
    "Neutral": 1,
    "Positive": 2
}

id2label = {
    0: "Negative",
    1: "Neutral",
    2: "Positive"
}

balanced_train_df["label"] = balanced_train_df["sentiment"].map(label2id)
val_df["label"] = val_df["sentiment"].map(label2id)
test_df["label"] = test_df["sentiment"].map(label2id)
```

```python
from transformers import AutoTokenizer

model_name = "distilbert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
from datasets import Dataset

train_dataset = Dataset.from_pandas(
    balanced_train_df[["review_text", "label"]],
    preserve_index=False
)

val_dataset = Dataset.from_pandas(
    val_df[["review_text", "label"]],
    preserve_index=False
)

test_dataset = Dataset.from_pandas(
    test_df[["review_text", "label"]],
    preserve_index=False
)
```

```python
def tokenize(batch):
    return tokenizer(
        batch["review_text"],
        truncation=True,
        padding="max_length",
        max_length=256
    )

train_dataset = train_dataset.map(tokenize, batched=True)
val_dataset = val_dataset.map(tokenize, batched=True)
test_dataset = test_dataset.map(tokenize, batched=True)
```

```python
train_dataset = train_dataset.remove_columns(["review_text"])
val_dataset = val_dataset.remove_columns(["review_text"])
test_dataset = test_dataset.remove_columns(["review_text"])

train_dataset.set_format("torch")
val_dataset.set_format("torch")
test_dataset.set_format("torch")
```

```python
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support
)

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)

    accuracy = accuracy_score(labels, predictions)

    precision, recall, f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="macro",
        zero_division=0
    )

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }
```

```python
from transformers import pipeline
import torch

hf_model_name = "SebasLopez-ai/distilbert-amazon-reviews-sentiment"

device = 0 if torch.cuda.is_available() else -1

hf_classifier = pipeline(
    "text-classification",
    model=hf_model_name,
    tokenizer=hf_model_name,
    device=device
)
```

```python
hf_results = hf_classifier(
    val_df["review_text"].tolist(),
    batch_size=32,
    truncation=True,
    max_length=256
)

hf_predictions = [result["label"] for result in hf_results]
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(
    val_df["sentiment"],
    hf_predictions,
    labels=["Negative", "Neutral", "Positive"]
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Negative", "Neutral", "Positive"]
)

disp.plot()
plt.show()
```
